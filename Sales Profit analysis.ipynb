```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
```

```python
df.columns
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["ship_date"] = pd.to_datetime(df["ship_date"])
```

```python
df.head()
```

```python
df.columns
```

```python
df.drop(columns=["order_dater"], inplace=True)
```

```python
df.columns
```

```python
df["delivery_days"] = (
    df["ship_date"] - df["order_date"]
).dt.days
```

```python
df.head()
```

```python
df["year"] = df["order_date"].dt.year
```

```python
df["month"] = df["order_date"].dt.month_name()
```

```python
df["month_no"] = df["order_date"].dt.month
```

```python
df["quarter"] = "Q" + df["order_date"].dt.quarter.astype(str)
```

```python
df.head()
```

```python
customer_orders = df.groupby("customer_id")["order_id"].nunique()
```

```python
customer_segment = {}

for customer, orders in customer_orders.items():

    if orders == 1:
        customer_segment[customer] = "New"

    elif orders <= 5:
        customer_segment[customer] = "Returning"

    else:
        customer_segment[customer] = "Loyal"
```

```python
df["customer_segment"] = df["customer_id"].map(customer_segment)
```

```python
df.head()
```

```python
df["discount_percentage"] = df["discount"] * 100
```

```python
df["is_profitable"] = df["profit"].apply(
    lambda x: "Profit" if x >= 0 else "Loss"
)
```

```python
def order_category(sales):
    if sales < 100:
        return "Low Value"
    elif sales < 500:
        return "Medium Value"
    else:
        return "High Value"

df["order_value_category"] = df["sales"].apply(order_category)
```

```python
df.head()
```

```python
df["is_profitable"] = df["profit"].apply(
    lambda x: "Profit" if x >= 0 else "Loss"
)
```

```python
df.head()
```

```python
df["profit_margin"] = round((df["profit"] / df["sales"]) * 100, 2)
```
